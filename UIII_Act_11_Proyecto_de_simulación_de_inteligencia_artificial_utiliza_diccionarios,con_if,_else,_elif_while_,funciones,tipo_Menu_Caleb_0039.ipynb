## Realizando una simulacion de inteligencia artificial utilizando diccionarios,con if, else, elif, while, funciones, tipo Menu
# Problema numero 15: Soporte Tecnico

```python
# Programa de Simulación de IA para Soportes Técnicos
# Basado en estructuras de datos de Python: Listas y Diccionarios [1, 2]

# 1. Iniciamos con una lista vacía para almacenar nuestros soportes [3]
soportes_tecnicos = []

def mostrar_menu():
    print("\n" + "="*50)
    print("   SISTEMA DE IA - GESTIÓN DE SOPORTES TÉCNICOS")
    print("="*50)
    print("1. Agregar nuevo Soporte Técnico")
    print("2. Ver listado de Soportes (Tabla)")
    print("3. Consultar Asistente de IA (Análisis de Datos)")
    print("4. Salir")
    print("="*50)

def agregar_soporte():
    # Creamos un diccionario vacío para el nuevo registro [4]
    soporte = {}

    print("\n--- Ingreso de Datos del Soporte ---")
    # Entrada de datos (Key: Value) [5]
    soporte["Sistema de tickets"] = input("Nombre del sistema de tickets: ")
    soporte["Nivel de soporte"] = input("Nivel (1, 2 o 3): ")
    soporte["Empresa"] = input("Nombre de la empresa: ")
    soporte["Tiempo de respuesta"] = input("Tiempo de respuesta (ej. 2 horas): ")
    soporte["Software de diagnóstico"] = input("Software de diagnóstico utilizado: ")
    soporte["Dispositivos gestionados"] = input("Cantidad de dispositivos gestionados: ")

    # Agregamos el diccionario a nuestra lista [3]
    soportes_tecnicos.append(soporte)
    print("\n✅ Soporte agregado exitosamente.")

def mostrar_tabla():
    if not soportes_tecnicos:
        print("\n⚠️ La lista está vacía actualmente.")
    else:
        print("\n" + "-"*130)
        # Encabezados alineados usando f-strings para una buena distribución [6]
        print(f"{'Tickets':<20} | {'Nivel':<8} | {'Empresa':<20} | {'Respuesta':<15} | {'Software':<20} | {'Dispositivos':<12}")
        print("-"*130)

        # Ciclo for para recorrer la lista de diccionarios [7]
        for s in soportes_tecnicos:
            print(f"{s['Sistema de tickets']:<20} | {s['Nivel de soporte']:<8} | {s['Empresa']:<20} | {s['Tiempo de respuesta']:<15} | {s['Software de diagnóstico']:<20} | {s['Dispositivos gestionados']:<12}")
        print("-"*130)

def simular_ia():
    # Simulación de lógica de IA usando condicionales if-elif-else [8, 9]
    if not soportes_tecnicos:
        print("\n🤖 IA: No tengo datos suficientes para analizar.")
    else:
        print("\n🤖 ANALIZADOR DE IA ACTIVADO...")
        for s in soportes_tecnicos:
            nivel = s["Nivel de soporte"]
            print(f"\n> Análisis para la empresa {s['Empresa']}:")

            # Lógica de toma de decisiones "IA"
            if nivel == "1":
                print("   Recomendación: Soporte básico detectado. Sugerir guías de auto-ayuda.")
            elif nivel == "2":
                print("   Recomendación: Soporte intermedio. Programar revisión remota inmediata.")
            elif nivel == "3":
                print("   Recomendación: ¡CRÍTICO! Escalar a ingenieros especializados senior.")
            else:
                print("   Aviso: Nivel de soporte no reconocido. Verificar entrada de datos.")

# 2. Bucle principal del programa (while True) [10, 11]
while True:
    mostrar_menu()
    opcion = input("Selecciona una opción (1-4): ")

    if opcion == "1":
        agregar_soporte()
    elif opcion == "2":
        mostrar_tabla()
    elif opcion == "3":
        simular_ia()
    elif opcion == "4":
        print("\n🤖 IA: Finalizando sesión. ¡Hasta pronto!")
        break # Sale del ciclo True [11]
    else:
        print("\n❌ Opción no válida, intenta de nuevo.")
```
